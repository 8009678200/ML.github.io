Exploratory Data Analysis


Gender

```python
df.columns
```

```python
ax = sns.countplot(x = 'Gender', data = df)

for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Gender', y = 'Amount', data = sales_gen)
```

From above graph we can see that the most of the buyers are female even the purchasing power of female is greater than male.

Age

```python
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sale_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Age Group', y = 'Amount', data = sale_age)
```

From above graph we can see that the most of the buyers are the age group of 26-35

State

```python
# total no of orders from the top 10 states

sales_states = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x = 'State', y = 'Orders', data = sales_states)
```

```python
#total amount/sales from top 10 states
sales_states = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.barplot(x = 'State', y = 'Amount', data = sales_states)
```

"from above graph we can see that the the most of the orders unexpectively are from u.p, maharashtra, and karnataka respectively but  total sales/amount is fron u.p, maharastra,karnataka."

martial status

```python
ax = sns.countplot(data=df, x = 'Marital_Status')

sns.set(rc={'figure.figsize':(7,5)})
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sales_states = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(x = 'Marital_Status', y = 'Amount', hue='Gender', data = sales_states)
```

```python
sales_states = df.groupby('Marital_Status', as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(x = 'Marital_Status', y = 'Amount', data = sales_states)
```

"from above graph we can see that the most of the buyers are married".

Occupation

```python
ax = sns.countplot(data=df, x = 'Occupation')
sns.set(rc={'figure.figsize':(20,5)})

for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sales_occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x = 'Occupation', y = 'Amount', data = sales_occupation)
```

from above graph we can see that the most of the buyers are from are IT sector,healthcare and aviation
